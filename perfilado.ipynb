### Categorías raras

En el análisis de variables categóricas se identificaron patrones esperados en departamentos y provincias, con predominancia de registros en regiones como Áncash y Amazonas.

Sin embargo, se detectaron valores no esperados en algunas columnas, incluyendo enlaces web dentro de campos categóricos, lo que indica posibles errores de carga o contaminación del dataset.

Estas inconsistencias deben ser revisadas y depuradas para garantizar la correcta interpretación de las variables geográficas.

```python
def iqr_outliers(df, col):
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    li = Q1 - 1.5 * IQR
    ls = Q3 + 1.5 * IQR

    return df[(df[col] < li) | (df[col] > ls)]

num_cols = df3.select_dtypes(include=np.number).columns

for col in num_cols:
    print(col, "outliers:", len(iqr_outliers(df3, col)))
```

### Outliers

El análisis de outliers en el dataset RENAMU muestra que varias variables presentan valores extremos, especialmente aquellas asociadas a preguntas tipo encuesta (P11A, P11B, P17, entre otras).

Sin embargo, estos outliers no representan necesariamente errores, ya que muchas de estas variables corresponden a respuestas codificadas o categorías discretas propias del diseño del cuestionario.

En contraste, variables estructurales como códigos geográficos y administrativos presentan una baja o nula presencia de valores atípicos, lo que indica consistencia en la base de datos.

```python
inconsistencias = 0

for col in num_cols:
    inconsistencias += (df3[col] < 0).sum()

consistencia = 1 - inconsistencias / (df3.shape[0] * len(num_cols))
consistencia
```

### Consistencia de datos

El análisis de consistencia del dataset RENAMU muestra un valor de 1.0, lo que indica que no se encontraron valores inválidos dentro de las variables evaluadas, como datos negativos o fuera de rango lógico.

Dentro de estas columnas se observan valores correspondientes a empresas operadoras y fechas, los cuales presentan una distribución coherente, aunque requieren revisión y estandarización de nombres de columnas.

Estas inconsistencias estructurales deben ser corregidas para asegurar una correcta interpretación y análisis del dataset.

```python
import numpy as np

def iqr_outliers(df, col):
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    li = Q1 - 1.5 * IQR
    ls = Q3 + 1.5 * IQR

    return df[(df[col] < li) | (df[col] > ls)]

num_cols = df4.select_dtypes(include=np.number).columns

for col in num_cols:
    print(col, "outliers:", len(iqr_outliers(df4, col)))
```

### Outliers

El análisis de outliers en el dataset de quejas y reclamos muestra que la variable numérica evaluada no presenta valores atípicos significativos.

Esto indica que los datos se encuentran dentro de rangos esperados y no existen valores extremos que afecten el análisis estadístico de la información.

```python
inconsistencias = 0

for col in num_cols:
    inconsistencias += (df4[col] < 0).sum()

consistencia = 1 - inconsistencias / (df4.shape[0] * len(num_cols))
consistencia
```
