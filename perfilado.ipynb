## Outliers

El dataset presenta consistencia en nombres geográficos, categorías y tipos de datos.
Las variables porcentuales y geográficas mantienen rangos válidos y coherentes con el contexto del análisis.

```python
if completitud >= 95 and unicidad >= 95:
    calidad = "Alta"
elif completitud >= 80:
    calidad = "Media"
else:
    calidad = "Baja"

print("Clasificación de calidad:", calidad)
```

### Categorías raras

Se identificaron pequeñas inconsistencias en variables categóricas como empresa, ubicación y tecnología (diferencias de escritura, mayúsculas o espacios).  
Estas deben normalizarse para evitar duplicidad en el análisis.

```python
Q1 = df2['CANT_CONEXIONES'].quantile(0.25)
Q3 = df2['CANT_CONEXIONES'].quantile(0.75)
IQR = Q3 - Q1

li = Q1 - 1.5 * IQR
ls = Q3 + 1.5 * IQR

outliers = df2[(df2['CANT_CONEXIONES'] < li) | (df2['CANT_CONEXIONES'] > ls)]

len(outliers)
```

### Outliers

Se detectaron valores extremos en variables numéricas, principalmente en la cantidad de conexiones.  
Sin embargo, estos valores reflejan diferencias reales entre distritos y no necesariamente errores en los datos.

```python
completitud = 100 - (
    df2.isnull().sum().sum()
    / (df2.shape[0] * df2.shape[1])
) * 100

print("Completitud:", round(completitud, 2), "%")
```

```python
unicidad = 100 - (
    df2.duplicated().sum()
    / len(df2)
) * 100

print("Unicidad:", round(unicidad, 2), "%")
```

### Consistencia de datos

Se evaluó la consistencia del dataset verificando la presencia de valores inválidos en las variables numéricas, como valores negativos o fuera de rango lógico.

Los resultados indican un nivel adecuado de consistencia, sin errores significativos que afecten la integridad del análisis.

```python
if completitud >= 95 and unicidad >= 95:
    calidad = "Alta"
elif completitud >= 80:
    calidad = "Media"
else:
    calidad = "Baja"

print("Clasificación de calidad:", calidad)
```

### Conclusión del Dataset Conexiones de Servicio

El dataset presenta una estructura adecuada para el análisis de la conectividad de internet en el Perú, permitiendo el estudio por distrito, empresa operadora y rangos de velocidad.

Se identificaron algunas inconsistencias menores en variables categóricas, las cuales pueden ser normalizadas para mejorar la calidad del análisis. En cuanto a valores numéricos, los outliers detectados reflejan principalmente diferencias reales en la distribución del servicio y no errores en los datos.

En general, el dataset es consistente, con buena completitud y unicidad, siendo apto para el análisis de la brecha de acceso a internet.
